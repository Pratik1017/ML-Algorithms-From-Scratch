```python
class MSGDRegressor:

  def __init__(self,Lr, epochs,batch_size):
    self.m = None
    self.b = None
    self.Lr = Lr
    self.epochs = epochs
    self.batch_size = batch_size

  def fit(self,X_train, y_train):

    self.b = 0
    self.m = np.ones(X_train.shape[1])

    for i in range(self.epochs):
      for j in range(X_train.shape[0] // self.batch_size):

        idx = np.random.randint(0, X_train.shape[0], self.batch_size)

        y_hat = np.dot(X_train[idx] , self.m) + self.b
        loss_slope_b = -2 * np.mean(y_train[idx] - y_hat)
        self.b = self.b - (self.Lr * loss_slope_b)

        loss_slope_m = -2 * np.dot((y_train[idx] - y_hat) , X_train[idx])
        self.m = self.m - (self.Lr * loss_slope_m)

    print(self.m , self.b)
```

```python
SGD = MSGDRegressor(0.001 , 6,3)
```
